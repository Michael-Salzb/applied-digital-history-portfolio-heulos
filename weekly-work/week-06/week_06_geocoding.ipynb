```python

import pandas as pd

# ============================================================
# 1. Check the original dataset
# ============================================================

print("1. ORIGINAL DATASET")

original_rows = len(books_coordinates)

unique_editions = books_coordinates["edition_id"].nunique(
    dropna=True
)

missing_ids = books_coordinates["edition_id"].isna().sum()
duplicate_ids = books_coordinates["edition_id"].duplicated().sum()

print("Total edition records:", original_rows)
print("Unique edition_id values:", unique_editions)
print("Missing edition_id values:", missing_ids)
print("Duplicated edition_id values:", duplicate_ids)

# Every record should have its own edition ID.
assert missing_ids == 0, "Some edition IDs are missing!"
assert duplicate_ids == 0, "Some edition IDs are duplicated!"

print("Edition IDs are unique: YES")


# ============================================================
# 2. Identify records with usable coordinates
# ============================================================

print("\n2. COORDINATE VALIDATION")

# Check that both coordinate columns contain numbers.
latitude = pd.to_numeric(
    books_coordinates["latitude"], errors="coerce"
)

longitude = pd.to_numeric(
    books_coordinates["longitude"], errors="coerce"
)

# A coordinate is usable only when both numbers are
# present and inside the valid geographic ranges.
usable = (
    latitude.notna()
    & longitude.notna()
    & latitude.between(-90, 90)
    & longitude.between(-180, 180)
    & (books_coordinates["coordinate_parse_status"] == "success")
)

usable_rows = int(usable.sum())
excluded_rows = int((~usable).sum())

print("Records with usable coordinates:", usable_rows)
print("Records excluded from mapping:", excluded_rows)


# ============================================================
# 3. Inspect records excluded from mapping
# ============================================================

print("\n3. EXCLUDED RECORDS")

excluded_records = books_coordinates.loc[
    ~usable,
    ["place_interpreted", "match_status"]
]

# Display each distinct combination once.
excluded_values = (
    excluded_records
    .drop_duplicates()
    .sort_values(
        ["place_interpreted", "match_status"],
        na_position="last"
    )
)

print("Distinct place/status combinations excluded:",
      len(excluded_values))

display(excluded_values)

print("These records are excluded only from the map table.")
print("They remain in books_coordinates.")


# ============================================================
# 4. Create a temporary table of usable records
# ============================================================

# Copy the usable records so the original stays unchanged.
valid_records = books_coordinates.loc[
    usable,
    ["edition_id", "place_interpreted"]
].copy()

# Add the validated numerical coordinates.
valid_records["latitude"] = latitude.loc[usable]
valid_records["longitude"] = longitude.loc[usable]

# Grouping requires a place name as well as coordinates.
# Keep missing place names as a separate group.
# This prevents losing otherwise usable records.


# ============================================================
# 5. Identify places with multiple coordinate pairs
# ============================================================

print("\n4. PLACES WITH MULTIPLE COORDINATE PAIRS")

# Keep distinct place/coordinate combinations.
place_coordinates = valid_records[
    ["place_interpreted", "latitude", "longitude"]
].drop_duplicates()

# Count coordinate pairs assigned to each place.
coordinate_counts = (
    place_coordinates
    .groupby("place_interpreted", dropna=False)
    .size()
    .reset_index(name="coordinate_pair_count")
)

# Select places with more than one coordinate pair.
multiple_coordinates = coordinate_counts[
    coordinate_counts["coordinate_pair_count"] > 1
]

print("Places assigned multiple coordinate pairs:",
      len(multiple_coordinates))

if len(multiple_coordinates) > 0:

    # Show each conflicting coordinate pair.
    conflicting_pairs = place_coordinates.merge(
        multiple_coordinates[["place_interpreted"]],
        on="place_interpreted",
        how="inner"
    )

    display(
        conflicting_pairs.sort_values(
            ["place_interpreted", "latitude", "longitude"],
            na_position="last"
        )
    )

else:
    print("No places with multiple coordinate pairs found.")


# ============================================================
# 6. Aggregate editions by place and coordinate pair
# ============================================================

print("\n5. AGGREGATED PRINTING PLACES")

city_counts = (
    valid_records
    .groupby(
        ["place_interpreted", "latitude", "longitude"],
        dropna=False
    )
    .agg(editions=("edition_id", "count"))
    .reset_index()
)

# Sort from the most editions to the fewest.
city_counts = city_counts.sort_values(
    "editions",
    ascending=False
).reset_index(drop=True)

print("Number of place/coordinate combinations:",
      len(city_counts))

display(city_counts)


# ============================================================
# 7. Validate the aggregated totals
# ============================================================

print("\n6. AGGREGATION VALIDATION")

total_editions = int(city_counts["editions"].sum())

print("Records with usable coordinates:", usable_rows)
print("Sum of editions in city_counts:", total_editions)
print("Excluded records:", excluded_rows)

assert total_editions == usable_rows, (
    "Aggregation totals do not match!"
)

assert usable_rows + excluded_rows == original_rows, (
    "Some records were not accounted for!"
)

# Confirm that each combination occurs only once.
assert not city_counts.duplicated(
    subset=["place_interpreted", "latitude", "longitude"]
).any(), "Duplicate place/coordinate combinations found!"

# Confirm that the source DataFrame retained all rows.
assert len(books_coordinates) == original_rows

print("Aggregation totals match: YES")
print("All original records accounted for: YES")
print("Unique place/coordinate combinations: YES")


# ============================================================
# 8. Save the aggregated CSV
# ============================================================

output_filename = "printing_places_aggregated.csv"

city_counts.to_csv(
    output_filename,
    index=False,
    encoding="utf-8-sig"
)

print("\n7. FILE SAVED")
print("Filename:", output_filename)
print("Rows in exported CSV:", len(city_counts))
print("Original books_coordinates remains unchanged.")
```

```python

# ============================================================
# Interactive map of historical printing places
# Input: city_counts
# Output: printing_places_map.html
# ============================================================

import folium
import numpy as np
import pandas as pd
from IPython.display import display
from html import escape

# ------------------------------------------------------------
# 1. Check the input data
# ------------------------------------------------------------

required_columns = [
    "place_interpreted",
    "latitude",
    "longitude",
    "editions"
]

for column in required_columns:
    assert column in city_counts.columns, (
        f"Missing column: {column}"
    )

print("Printing places in city_counts:", len(city_counts))

# Make a separate copy for mapping.
# The original city_counts remains unchanged.
map_data = city_counts.copy()

# Check numerical values without modifying city_counts.
for column in ["latitude", "longitude", "editions"]:
    map_data[column] = pd.to_numeric(
        map_data[column],
        errors="coerce"
    )

# Keep only rows that can be mapped.
valid = (
    map_data["latitude"].between(-90, 90)
    & map_data["longitude"].between(-180, 180)
    & map_data["editions"].notna()
    & (map_data["editions"] > 0)
)

map_data = map_data.loc[valid].copy()

print("Places with valid mapping data:", len(map_data))
print("Rows skipped:", len(city_counts) - len(map_data))


# ------------------------------------------------------------
# 2. Create the map without default tiles
# ------------------------------------------------------------

# Initial center: Central Europe
# Initial zoom: 5 (appropriate for the basemap)
printing_map = folium.Map(
    location=[49.5, 12.0],
    zoom_start=5,
    tiles=None,
    control_scale=True
)

# Add the requested custom basemap.
folium.TileLayer(
    tiles=(
        "https://tiles.openfreemap.org/"
        "natural_earth/ne2sr/{z}/{x}/{y}.png"
    ),
    name="OpenFreeMap Natural Earth",
    attr=(
        "Map tiles: OpenFreeMap | "
        "Map data: Natural Earth"
    ),
    max_zoom=6,
    overlay=False,
    control=True,
    show=True
).add_to(printing_map)


# ------------------------------------------------------------
# 3. Add one circle marker per printing place
# ------------------------------------------------------------

for _, row in map_data.iterrows():

    place = (
        str(row["place_interpreted"])
        if pd.notna(row["place_interpreted"])
        else "Unknown place"
    )

    latitude = float(row["latitude"])
    longitude = float(row["longitude"])
    editions = int(row["editions"])

    # Square-root scaling:
    # Larger edition counts produce larger circles.
    radius = 3 + 2.2 * np.sqrt(editions)

    # Escape place names for safe HTML display.
    safe_place = escape(place)

    # Short text appears when hovering over a circle.
    tooltip_text = f"{safe_place}: {editions} editions"

    # Detailed information appears when clicking.
    popup_text = f"""
    <div style="min-width: 190px;">
        <b>{safe_place}</b><br>
        Recorded editions: {editions}<br>
        Latitude: {latitude:.6f}<br>
        Longitude: {longitude:.6f}
    </div>
    """

    # Create one circle for this row.
    folium.CircleMarker(
        location=[latitude, longitude],
        radius=radius,
        color="#17365D",
        weight=1.5,
        opacity=0.9,
        fill=True,
        fill_color="#3274B8",
        fill_opacity=0.55,
        tooltip=folium.Tooltip(tooltip_text),
        popup=folium.Popup(popup_text, max_width=300)
    ).add_to(printing_map)


# ------------------------------------------------------------
# 4. Fit the map to the plotted coordinates
# ------------------------------------------------------------

if len(map_data) > 0:

    bounds = [
        [
            map_data["latitude"].min(),
            map_data["longitude"].min()
        ],
        [
            map_data["latitude"].max(),
            map_data["longitude"].max()
        ]
    ]

    # Adjust the viewport to contain all plotted places.
    # Do not zoom beyond the basemap's maximum zoom.
    printing_map.fit_bounds(
        bounds,
        padding=(30, 30),
        max_zoom=6
    )

else:
    print("No valid locations found; using Central Europe view.")


# ------------------------------------------------------------
# 5. Add a layer control
# ------------------------------------------------------------

folium.LayerControl().add_to(printing_map)


# ------------------------------------------------------------
# 6. Save the interactive HTML map
# ------------------------------------------------------------

output_file = "printing_places_map.html"

printing_map.save(output_file)

print("\nMap successfully created!")
print("File:", output_file)
print("Circle markers:", len(map_data))
print("Original city_counts unchanged.")

# ------------------------------------------------------------
# 7. Display the map in Google Colab
# ------------------------------------------------------------

display(printing_map)
```
